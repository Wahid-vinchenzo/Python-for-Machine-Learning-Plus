### **List**

### List methods
---

- `append()`: Adds an element at the end of the list
- `clear()`: Removes all the elements from the list
- `copy()`: Returns a copy of the list
- `count()`: Returns the number of elements with the specified value
- `extend()`: Add the elements of a list (or any iterable), to the end of the current list
- `index()`: Returns the index of the first element with the specified value
- `insert()`: Adds an element at the specified position
- `pop()`: Removes the element at the specified position
- `remove()`: Removes the item with the specified value
- `reverse()`: Reverses the order of the list
- `sort()`: Sorts the list

| কাজ                        | Method                    |
| -------------------------- | ------------------------- |
| Create list                | `[]`                      |
| Access item                | `list[index]`             |
| Add item শেষে              | `append()`                |
| Add item specific position | `insert()`                |
| Remove by value            | `remove()`                |
| Remove by index            | `pop()`                   |
| Remove all                 | `clear()`                 |
| Delete item/list           | `del`                     |
| Change item                | `list[index] = value`     |
| Join lists                 | `+`, `extend()`           |
| Loop                       | `for`                     |
| Sort                       | `sort()`                  |
| Reverse                    | `reverse()`               |
| Copy                       | `copy()`, `list()`, `[:]` |


````python
# =========================================
# 1. Creating Different Types of List
# =========================================

# Empty list
emp_list = []

# String list
str_list = ["i", "love", "python"]

# Integer list
int_list = [1, 2, 3, 4]

# Float list
float_list = [1.33, 5.55, 5.66]

# Mixed data type list
mix_list = ["i", 1, 1.55]

print(emp_list)
print(str_list)
print(int_list)
print(float_list)
print(mix_list)


# =========================================
# 2. Accessing List Elements
# =========================================

# Access first element
print(mix_list[0])

# Access last element using negative indexing
print(mix_list[-1])

# Access multiple elements using slicing
print(mix_list[1:3])


# =========================================
# 3. Operations on List Elements
# =========================================

# Mathematical operation (only works on numeric values)
print(mix_list[-1] + 2)

# String concatenation
print(mix_list[0] + " love you")


# =========================================
# 4. Adding Elements to List
# =========================================

# append() → adds value at the end
mix_list.append(100)
print(mix_list)

# insert(index, value) → adds value at specific position
mix_list.insert(0, 500)
print(mix_list)


# =========================================
# 5. Removing Elements from List
# =========================================

# remove(value) → removes specific value
mix_list.remove(1)
print(mix_list)

# pop(index) → removes element by index
thislist = ["apple", "banana", "cherry"]

thislist.pop(1)   # remove banana
print(thislist)

thislist.pop()    # remove last item
print(thislist)

# clear() → removes all elements
mix_list.clear()
print(mix_list)


# =========================================
# 6. Delete Elements Using del
# =========================================

mix_list = ["a", "b", "c", "d"]

# Delete specific index
del mix_list[0]
print(mix_list)

"""
# Delete entire list
del mix_list
print(mix_list)
"""


# =========================================
# 7. Updating / Changing List Items
# =========================================

thislist = ["apple", "banana", "cherry"]

# Change single item
thislist[1] = "blackcurrant"
print(thislist)

# Change multiple items using slicing
thislist = ["apple", "banana", "cherry", "orange", "kiwi", "mango"]

thislist[1:3] = ["blackcurrant", "watermelon"]
print(thislist)


# =========================================
# 8. Creating List from Tuple
# =========================================

my_tuple = (1, 2, 3, 4)

# Convert tuple to list
my_list = list(my_tuple)
print(my_list)


# =========================================
# 9. Joining / Combining Lists
# =========================================

list1 = [1, 2, 3]
list2 = [4, 5, 6]

# Using + operator
new_list = list1 + list2
print(new_list)

# Using extend()
thislist = ["apple", "banana", "cherry"]
tropical = ["mango", "pineapple", "papaya"]

thislist.extend(tropical)
print(thislist)

# extend() also works with tuple
thislist = ["apple", "banana", "cherry"]
thistuple = ("kiwi", "orange")

thislist.extend(thistuple)
print(thislist)


# =========================================
# 10. Looping Through List
# =========================================

thislist = ["apple", "banana", "cherry"]

# Normal loop
for x in thislist:
    print(x)

# Loop using index
for i in range(len(thislist)):
    print(thislist[i])

# List comprehension print
[print(x) for x in thislist]


# =========================================
# 11. Creating List Using range()
# =========================================

# Create list from 0 to 9
newlist = [x for x in range(10)]
print(newlist)

# Create list with condition
newlist = [x for x in range(10) if x < 5]
print(newlist)


# =========================================
# 12. Sorting List
# =========================================

# Sort strings alphabetically
thislist = ["orange", "mango", "kiwi", "pineapple", "banana"]
thislist.sort()
print(thislist)

# Sort numbers
thislist = [100, 50, 65, 82, 23]
thislist.sort()
print(thislist)

# Descending order sort
thislist = ["orange", "mango", "kiwi", "pineapple", "banana"]
thislist.sort(reverse=True)
print(thislist)


# =========================================
# 13. Reverse List
# =========================================

thislist = ["banana", "Orange", "Kiwi", "cherry"]

# Reverse original order
thislist.reverse()
print(thislist)


# =========================================
# 14. Custom Sorting
# =========================================

# Sort based on nearest value to 50
def myfunc(n):
    return abs(n - 50)

thislist = [100, 50, 65, 82, 23]
thislist.sort(key=myfunc)
print(thislist)


# =========================================
# 15. Copying a List
# =========================================

thislist = ["apple", "banana", "cherry"]

# Method 1: copy()
mylist = thislist.copy()
print(mylist)

# Method 2: list()
mylist = list(thislist)
print(mylist)

# Method 3: slicing
mylist = thislist[:]
print(mylist)
````

# **Tuple Method / Operation Cheat Sheet ⚡**

| কাজ               | Method / Syntax                 |
| ----------------- | ------------------------------- |

| Find length       | `len(tuple)`                    |

````python
# =========================================
# 1. Creating Tuple
# =========================================

# Normal tuple
thistuple = ("apple", "banana", "cherry")
print(thistuple)


# =========================================
# 2. Size of Tuple
# =========================================

# len() → returns total number of elements
print(len(thistuple))


# =========================================
# 3. Tuple With One Item
# =========================================

# Must use comma after single item
thistuple = ("apple",)
print(type(thistuple))   # tuple

# Without comma → it becomes string
thistuple = ("apple")
print(type(thistuple))   # str


# =========================================
# 4. Different Types of Tuple
# =========================================

# String tuple
tuple1 = ("apple", "banana", "cherry")

# Integer tuple
tuple2 = (1, 5, 7, 9, 3)

# Boolean tuple
tuple3 = (True, False, False)

print(tuple1)
print(tuple2)
print(tuple3)


# =========================================
# 5. Creating Tuple from Another Tuple
# =========================================

tuple4 = tuple1
print(tuple4)

# Concatenating multiple tuples
new_tuple = tuple2 + tuple1 + tuple3
print(new_tuple)


# =========================================
# 6. Accessing Tuple Elements
# =========================================

# Access specific index
print(new_tuple[1])

# Access all values using slicing
print(new_tuple[:])

# Access last element
print(new_tuple[-1])


# =========================================
# 7. Deleting Tuple
# =========================================

temp_tuple = ("a", "b", "c")

# Delete entire tuple
del temp_tuple

# print(temp_tuple) → Error হবে কারণ tuple delete হয়ে গেছে


# =========================================
# 8. Unpacking Tuple
# =========================================

fruits = ("apple", "banana", "cherry")

(green, yellow, red) = fruits

print(green)
print(yellow)
print(red)


# =========================================
# 9. Using * for Multiple Values
# =========================================

# Remaining values stored as list
(green, yellow, *red) = fruits

print(green)
print(yellow)
print(red)


# =========================================
# 10. * in Middle Position
# =========================================

fruits = ("apple", "mango", "papaya", "pineapple", "cherry")

(green, *tropic, red) = fruits

print(green)
print(tropic)
print(red)


# =========================================
# 11. Multiplying Tuple
# =========================================

fruits = ("apple", "banana", "cherry")

# Repeat tuple 2 times
mytuple = fruits * 2
print(mytuple)


# =========================================
# 12. count() Method
# =========================================

thistuple = (1, 3, 7, 8, 7, 5, 4, 6, 8, 5)

# Count how many times 5 appears
x = thistuple.count(5)
print(x)


# =========================================
# 13. index() Method
# =========================================

thistuple = (1, 3, 7, 8, 7, 5, 4, 6, 8, 5)

# Returns first index of value 8
x = thistuple.index(8)
print(x)


# =========================================
# 14. Add/Remove Items in Tuple
# =========================================

# Tuple is immutable (can't directly add/remove)

thistuple = ("apple", "banana", "cherry")

# Convert tuple → list
temp_list = list(thistuple)

# Add item
temp_list.append("orange")

# Remove item
temp_list.remove("banana")

# Convert list → tuple
thistuple = tuple(temp_list)

print(thistuple)
````

````python
# =========================================
# 1. Creating Set
# =========================================

# Normal set
myset = {"apple", "banana", "cherry"}
print(myset)


# =========================================
# 2. Different Types of Set
# =========================================

# String set
set1 = {"apple", "banana", "cherry"}

# Integer set
set2 = {1, 5, 7, 9, 3}

# Boolean set
set3 = {True, False}

# Mixed set
set4 = {"apple", True, 1, 2}

print(set1)
print(set2)
print(set3)
print(set4)


# =========================================
# 3. Creating Set from Tuple
# =========================================

# Convert tuple → set
thisset = set(("apple", "banana", "cherry"))
print(thisset)


# =========================================
# 4. Duplicate Values Not Allowed
# =========================================

thisset = {"apple", "banana", "cherry", "apple"}
print(thisset)   # duplicate remove হয়ে যাবে automatically


# =========================================
# 5. True = 1 and False = 0
# =========================================

thisset = {"apple", True, 1, 2}
print(thisset)

thisset = {"apple", False, True, 0}
print(thisset)


# =========================================
# 6. Length of Set
# =========================================

print(len(thisset))


# =========================================
# 7. Accessing Set Items
# =========================================

# Set indexing support করে না
# But loop দিয়ে access করা যায়

thisset = {"apple", "banana", "cherry"}

for x in thisset:
    print(x)


# =========================================
# 8. Check Item Exists
# =========================================

print("banana" in thisset)
print("banana" not in thisset)


# =========================================
# 9. Add Items
# =========================================

thisset = {"apple", "banana", "cherry"}

# Add single item
thisset.add("orange")
print(thisset)


# =========================================
# 10. Update Set (Add Multiple Items)
# =========================================

thisset = {"apple", "banana", "cherry"}
tropical = {"pineapple", "mango", "papaya"}

# Add another set
thisset.update(tropical)
print(thisset)


# Add list into set
thisset = {"apple", "banana", "cherry"}
mylist = ["kiwi", "orange"]

thisset.update(mylist)
print(thisset)


# =========================================
# 11. Remove Items
# =========================================

thisset = {"apple", "banana", "cherry"}

# remove() → error দিবে যদি item না থাকে
thisset.remove("banana")
print(thisset)


# discard() → error দিবে না
thisset = {"apple", "banana", "cherry"}
thisset.discard("banana")
print(thisset)


# pop() → random item remove করে
thisset = {"apple", "banana", "cherry"}

x = thisset.pop()
print(x)
print(thisset)


# clear() → সব remove
thisset.clear()
print(thisset)


# del → পুরো set delete
thisset = {"apple", "banana"}
del thisset


# =========================================
# 12. Union (Join Sets)
# =========================================

set1 = {"a", "b", "c"}
set2 = {1, 2, 3}

# union()
set3 = set1.union(set2)
print(set3)

# |
set3 = set1 | set2
print(set3)


# Multiple union
set3 = {"John", "Elena"}
set4 = {"apple", "banana"}

myset = set1.union(set2, set3, set4)
print(myset)


# =========================================
# 13. update()
# =========================================

set1 = {"a", "b", "c"}
set2 = {1, 2, 3}

# modifies original set
set1.update(set2)
print(set1)


# =========================================
# 14. Intersection
# Common values only
# =========================================

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

# intersection()
set3 = set1.intersection(set2)
print(set3)

# &
set3 = set1 & set2
print(set3)


# intersection_update()
set1.intersection_update(set2)
print(set1)


# =========================================
# 15. Difference
# First set unique values
# =========================================

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

# difference()
set3 = set1.difference(set2)
print(set3)

# -
set3 = set1 - set2
print(set3)


# difference_update()
set1.difference_update(set2)
print(set1)


# =========================================
# 16. Symmetric Difference
# Non-common values only
# =========================================

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

# symmetric_difference()
set3 = set1.symmetric_difference(set2)
print(set3)

# ^
set3 = set1 ^ set2
print(set3)


# symmetric_difference_update()
set1.symmetric_difference_update(set2)
print(set1)


# =========================================
# 17. Frozen Set
# =========================================

# Immutable set
x = frozenset({"apple", "banana", "cherry"})

print(x)
print(type(x))
````

### **Python Standard Library**

### **1. deque (Double Ended Queue)**

দুই পাশ থেকেই fast add/remove করা যায়।

````python
from collections import deque

# create deque
d = deque([1, 2, 3])

print(d)

# add right side
d.append(4)
print(d)

# add left side
d.appendleft(0)
print(d)

# remove right side
d.pop()
print(d)

# remove left side
d.popleft()
print(d)

# reverse deque
d.reverse()
print(d)

# count value
print(d.count(2))
````

**deque Methods Table**
---
| Method         | কাজ            |
| -------------- | -------------- |
| `append()`     | right side add |
| `appendleft()` | left side add  |
| `pop()`        | right remove   |
| `popleft()`    | left remove    |
| `reverse()`    | reverse        |
| `count()`      | count value    |


# **2. Counter**

Frequency counting এর জন্য use হয়।

````python
from collections import Counter

# count characters
x = Counter("banana")
print(x)

# count list values
nums = Counter([1, 2, 2, 3, 3, 3])
print(nums)

# most common values
print(nums.most_common(2))

# update counter
nums.update([2,2,2])
print(nums)
````

| Method          | কাজ                 |
| --------------- | ------------------- |
| `Counter()`     | create counter      |
| `most_common()` | most frequent items |
| `update()`      | add more frequency  |
| `elements()`    | return all elements |


# **3. defaultdict**

Missing key error avoid করে default value দেয়।

````python
from collections import defaultdict

# default int value = 0
d = defaultdict(int)

d["a"] += 1
d["b"] += 2

print(d)

# default list
students = defaultdict(list)

students["python"].append("Sakib")
students["python"].append("Rahim")

print(students)
````

| Method              | কাজ                  |
| ------------------- | -------------------- |
| `defaultdict(int)`  | default 0            |
| `defaultdict(list)` | default empty list   |
| `defaultdict(str)`  | default empty string |


# **4. namedtuple**

Tuple এর মতো but named field থাকে।

````python
from collections import namedtuple

# create namedtuple
Student = namedtuple("Student", ["name", "age", "dept"])

# create object
s1 = Student("Sakib", 22, "CSE")

print(s1)

# access by field name
print(s1.name)
print(s1.age)

# access by index
print(s1[0])
````

| Method         | কাজ                |
| -------------- | ------------------ |
| `namedtuple()` | create named tuple |
| `.field_name`  | access value       |
| indexing       | access like tuple  |


# Python Collections Guide 🚀

Python-এ data store করার জন্য সবচেয়ে common built-in containers হলো:

* **List**
* **Tuple**
* **Set**
* **Dictionary**

এগুলা অনেক time interview, problem solving, projects, database handling, API response processing—সব জায়গায় use হয়।

---

# 1. List `[]`

List হলো **ordered, mutable collection**

মানে:

* Order maintain করে
* Duplicate allow করে
* Value change করা যায়
* Different data types রাখতে পারে

```python
my_list = [1, 2, 3, "python", 5.5]

print(my_list)
```

### Features

✅ Ordered
✅ Mutable
✅ Duplicate allowed
✅ Indexing supported

---

### Common Methods

```python
append()
insert()
remove()
pop()
sort()
reverse()
clear()
copy()
extend()
```

---

### কখন List use হয়?

Use list when:

* Multiple values sequentially store করতে হবে
* Values frequently add/remove করতে হবে
* Indexing দরকার
* Ordered data দরকার

### Example:

* Student marks list
* Shopping cart items
* Todo list
* API response data

---

# 2. Tuple `()`

Tuple হলো **ordered, immutable collection**

মানে:

* Order maintain করে
* Duplicate allow করে
* Value change করা যায় না

```python
my_tuple = (1, 2, 3)

print(my_tuple)
```

---

### Features

✅ Ordered
❌ Mutable না
✅ Duplicate allowed
✅ Indexing supported

---

### Common Methods

```python
count()
index()
```

---

### কখন Tuple use হয়?

Use tuple when:

* Fixed data store করতে হবে
* Data accidental change prevent করতে হবে
* Faster performance দরকার

### Example:

* Coordinates `(x,y)`
* Database records
* RGB color values

```python
location = (23.7, 90.4)
```

---

# 3. Set `{}`

Set হলো **unordered, mutable collection of unique values**

মানে:

* Order guarantee নেই
* Duplicate allow করে না
* Unique values store করে

```python
my_set = {1, 2, 3, 4}

print(my_set)
```

---

### Features

❌ Ordered না
✅ Mutable
❌ Duplicate allowed না
❌ Indexing support নেই

---

### Common Methods

```python
add()
remove()
discard()
union()
intersection()
difference()
```

---

### কখন Set use হয়?

Use set when:

* Unique data দরকার
* Duplicate remove করতে হবে
* Fast membership checking দরকার

### Example:

* Unique usernames
* Unique tags
* Duplicate removal

```python
nums = [1,1,2,2,3]
print(set(nums))
```

---

# 4. Dictionary `{key:value}`

Dictionary হলো **key-value pair based collection**

```python
student = {
    "name": "Sakib",
    "age": 22
}
```

---

### Features

✅ Ordered (Python 3.7+)
✅ Mutable
❌ Duplicate keys allowed না
❌ Index based না

---

### Common Methods

```python
get()
keys()
values()
items()
update()
pop()
copy()
```

---

### কখন Dictionary use হয়?

Use dictionary when:

* Key দিয়ে data access করতে হবে
* Structured data store করতে হবে
* Real world object represent করতে হবে

### Example:

* User profile
* JSON data
* API response
* Database row

```python
user = {
    "name": "Rahim",
    "email": "rahim@gmail.com"
}
```

---

# List vs Tuple vs Set vs Dictionary

| Feature    | List | Tuple | Set  | Dictionary    |
| ---------- | ---- | ----- | ---- | ------------- |
| Syntax     | `[]` | `()`  | `{}` | `{key:value}` |
| Ordered    | ✅    | ✅     | ❌    | ✅             |
| Mutable    | ✅    | ❌     | ✅    | ✅             |
| Duplicates | ✅    | ✅     | ❌    | Keys ❌        |
| Indexing   | ✅    | ✅     | ❌    | ❌             |
| Key-Value  | ❌    | ❌     | ❌    | ✅             |

---

# কখন কোনটা use করবো?

### Use List → when order + modification needed

```python

```

---

### Use Tuple → when fixed data needed

```python

```

---

### Use Set → when unique values needed

```python

```

---

### Use Dictionary → when key-value mapping needed

```python
student = {
    "name": "Sakib",
    "id": 101
}
```
