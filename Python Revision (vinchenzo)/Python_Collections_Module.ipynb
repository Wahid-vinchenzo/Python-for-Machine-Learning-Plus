# 1️⃣ `Counter` — Counting Made Effortless

## 📖 What is `Counter`?

`Counter` is a **dict subclass** designed specifically for counting hashable objects.  
Elements are stored as dictionary keys, and their counts are stored as values.

**Syntax:**
```python
Counter(iterable_or_mapping)
```

### Why not just use a regular dict?
```python
# Manual approach — verbose!
text = "banana"
counts = {}
for char in text:
    counts[char] = counts.get(char, 0) + 1

# Counter approach — one line!
counts = Counter(text)
```

````python
# ============================================================
# COUNTER — CREATING COUNTERS
# ============================================================

# 1. From a string
char_count = Counter("banana")
print("From string:", char_count)

# 2. From a list
grades = Counter(["A", "B", "A", "C", "B", "A", "D", "B"])
print("From list:", grades)

# 3. From keyword arguments
inventory = Counter(apples=5, oranges=3, bananas=8)
print("From kwargs:", inventory)

# 4. From a dictionary
scores = Counter({"Alice": 100, "Bob": 85, "Carol": 92})
print("From dict:", scores)
````

````python
# ============================================================
# COUNTER — KEY METHODS
# ============================================================

words = Counter(["cat", "dog", "cat", "bird", "dog", "cat", "fish", "dog", "bird"])
print("Word counts:", words)

# most_common(n) → returns n most frequent elements
print("\nTop 2 most common:", words.most_common(2))
print("All sorted by frequency:", words.most_common())

# elements() → returns an iterator repeating each element by its count
print("\nelements():", list(words.elements()))

# Accessing a missing key returns 0 (not KeyError!)
print("\nCount of 'snake':", words["snake"])  # → 0

# total() → total count of all elements (Python 3.10+)
# print("Total count:", words.total())
````

````python
# ============================================================
# COUNTER — ARITHMETIC OPERATIONS
# ============================================================

inventory_day1 = Counter(apples=10, oranges=5, bananas=8)
sold_today     = Counter(apples=3, oranges=5, bananas=1)

# Subtraction — removes zero and negative counts
remaining = inventory_day1 - sold_today
print("Remaining stock:", remaining)

# Addition — combines two counters
restock = Counter(apples=20, grapes=15)
new_inventory = remaining + restock
print("After restock:", new_inventory)

# Intersection (&) → minimum of counts
c1 = Counter(a=5, b=3, c=1)
c2 = Counter(a=2, b=4, d=6)
print("Intersection (&):", c1 & c2)

# Union (|) → maximum of counts
print("Union (|):", c1 | c2)
````

````python
# ============================================================
# COUNTER — REAL-WORLD ML EXAMPLE: Word Frequency for NLP
# ============================================================

corpus = """
machine learning is a subset of artificial intelligence.
deep learning is a subset of machine learning.
neural networks power deep learning models.
machine learning models learn from data.
"""

# Tokenize and count
import re
tokens = re.findall(r'\b[a-z]+\b', corpus.lower())
word_freq = Counter(tokens)

print("Top 10 most frequent words:")
for word, count in word_freq.most_common(10):
    bar = "█" * count
    print(f"  {word:<15} {bar} ({count})")
````

# 2️⃣ `defaultdict` — Never Get a `KeyError` Again

## 📖 What is `defaultdict`?

`defaultdict` is a dict subclass that **automatically creates a default value** when accessing a missing key, instead of raising a `KeyError`.

**Syntax:**
```python
defaultdict(default_factory)
```

| `default_factory` | Default value when key is missing |
|-------------------|----------------------------------|
| `int`             | `0`                              |
| `list`            | `[]`                             |
| `set`             | `set()`                          |
| `str`             | `''`                             |
| `lambda: value`   | Any custom value                 |

````python
# ============================================================
# DEFAULTDICT — THE PROBLEM IT SOLVES
# ============================================================

# PROBLEM: Regular dict raises KeyError for missing keys
regular_dict = {}
try:
    regular_dict["missing_key"] += 1
except KeyError as e:
    print(f"❌ Regular dict raises: KeyError {e}")

# SOLUTION: defaultdict handles it gracefully
count_dict = defaultdict(int)
count_dict["missing_key"] += 1   # Starts at 0, then adds 1
print(f"✅ defaultdict(int): {count_dict['missing_key']}")


# defaultdict(list) — auto-creates empty list
student_courses = defaultdict(list)
student_courses["Alice"].append("Math")
student_courses["Alice"].append("Physics")
student_courses["Bob"].append("Chemistry")
print("\nStudent courses:", dict(student_courses))


# defaultdict(set) — auto-creates empty set
category_tags = defaultdict(set)
category_tags["ML"].add("supervised")
category_tags["ML"].add("regression")
category_tags["ML"].add("supervised")  # duplicate — ignored
category_tags["DL"].add("neural_networks")
print("Category tags:", dict(category_tags))
````

````python
# ============================================================
# DEFAULTDICT — GROUPING DATA (Very common pattern!)
# ============================================================

employees = [
    ("Alice", "Engineering"),
    ("Bob", "Marketing"),
    ("Carol", "Engineering"),
    ("Dave", "Marketing"),
    ("Eve", "Engineering"),
    ("Frank", "HR"),
]

# Group employees by department
by_dept = defaultdict(list)
for name, dept in employees:
    by_dept[dept].append(name)

print("Employees by department:")
for dept, names in by_dept.items():
    print(f"  {dept}: {names}")
````

````python
# ============================================================
# DEFAULTDICT — CUSTOM DEFAULT WITH LAMBDA
# ============================================================

# Default to a specific value
config = defaultdict(lambda: "NOT_SET")
config["debug"] = True
config["version"] = "1.0"

print("debug:", config["debug"])
print("version:", config["version"])
print("timeout:", config["timeout"])  # Not set → returns "NOT_SET"
print("log_level:", config["log_level"])  # Not set → returns "NOT_SET"


# Nested defaultdict (graph adjacency list)
graph = defaultdict(list)
graph["A"].extend(["B", "C"])
graph["B"].append("D")
graph["C"].extend(["D", "E"])

print("\nGraph adjacency list:")
for node, neighbors in graph.items():
    print(f"  {node} → {neighbors}")
````

````python
# ============================================================
# DEFAULTDICT — ML EXAMPLE: Building an Inverted Index
# ============================================================

documents = {
    1: "machine learning algorithms are powerful",
    2: "deep learning uses neural networks",
    3: "machine learning and deep learning overlap",
    4: "neural networks are the backbone of AI",
}

# Build inverted index: word → list of document IDs
inverted_index = defaultdict(list)

for doc_id, text in documents.items():
    for word in text.split():
        if doc_id not in inverted_index[word]:
            inverted_index[word].append(doc_id)

print("Inverted Index:")
for word in sorted(inverted_index):
    print(f"  '{word}': docs {inverted_index[word]}")

# Query: which docs contain both 'machine' and 'learning'?
query_words = ["machine", "learning"]
result_sets = [set(inverted_index[w]) for w in query_words]
common_docs = set.intersection(*result_sets)
print(f"\nDocs containing {query_words}: {sorted(common_docs)}")
````

# 3️⃣ `OrderedDict` — Order-Aware Dictionary

## 📖 What is `OrderedDict`?

An `OrderedDict` is a dict that **remembers the insertion order** of keys — and provides extra order-aware operations.

> ⚠️ **Note:** Since Python 3.7+, regular dicts also maintain insertion order. However, `OrderedDict` still provides:
> - `move_to_end()` method
> - Order-sensitive equality comparisons
> - `popitem(last=True/False)` control

````python
# ============================================================
# ORDEREDDICT — BASICS
# ============================================================

# Creating an OrderedDict
od = OrderedDict()
od["banana"] = 3
od["apple"] = 5
od["cherry"] = 1
od["date"] = 7

print("OrderedDict:", od)
print("Keys in insertion order:", list(od.keys()))

# Move to end
od.move_to_end("banana")
print("\nAfter move_to_end('banana'):", list(od.keys()))

# Move to beginning (last=False)
od.move_to_end("cherry", last=False)
print("After move_to_end('cherry', last=False):", list(od.keys()))
````

````python
# ============================================================
# ORDEREDDICT — ORDER-SENSITIVE EQUALITY
# ============================================================

# For regular dicts, order doesn't matter for equality
d1 = {"a": 1, "b": 2}
d2 = {"b": 2, "a": 1}
print("Regular dict equality (d1 == d2):", d1 == d2)  # True

# For OrderedDict, order DOES matter
od1 = OrderedDict([("a", 1), ("b", 2)])
od2 = OrderedDict([("b", 2), ("a", 1)])
print("OrderedDict equality (same order):", od1 == od1)   # True
print("OrderedDict equality (diff order):", od1 == od2)   # False

# popitem() with control
od = OrderedDict(a=1, b=2, c=3, d=4)
print("\npopitem(last=True):", od.popitem(last=True))    # removes 'd'
print("popitem(last=False):", od.popitem(last=False))   # removes 'a'
print("Remaining:", od)
````

````python
# ============================================================
# ORDEREDDICT — REAL-WORLD EXAMPLE: LRU Cache Implementation
# ============================================================

class LRUCache:
    """Least Recently Used (LRU) Cache using OrderedDict."""

    def __init__(self, capacity: int):
        self.capacity = capacity
        self.cache = OrderedDict()

    def get(self, key):
        if key not in self.cache:
            return -1
        # Move accessed item to end (most recently used)
        self.cache.move_to_end(key)
        return self.cache[key]

    def put(self, key, value):
        if key in self.cache:
            self.cache.move_to_end(key)
        self.cache[key] = value
        # If over capacity, remove least recently used (front)
        if len(self.cache) > self.capacity:
            evicted = self.cache.popitem(last=False)
            print(f"  [EVICTED] {evicted}")

    def __repr__(self):
        return f"LRUCache({dict(self.cache)})"


# Demo
cache = LRUCache(capacity=3)
cache.put("model_v1", "weights_A")
cache.put("model_v2", "weights_B")
cache.put("model_v3", "weights_C")
print("Cache:", cache)

cache.get("model_v1")  # Access model_v1 → moves to end
print("After accessing model_v1:", cache)

print("Adding model_v4 (capacity=3, so oldest evicted):")
cache.put("model_v4", "weights_D")
print("Cache:", cache)
````

# 4️⃣ `deque` — Double-Ended Queue

## 📖 What is `deque`?

A `deque` (pronounced *deck*) is a **double-ended queue** — you can add and remove elements from **both ends** in **O(1)** time.

| Operation | `list` | `deque` |
|-----------|--------|---------|
| Append to right | O(1) | O(1) |
| Append to left | O(n) | O(1) ✅ |
| Pop from right | O(1) | O(1) |
| Pop from left | O(n) | O(1) ✅ |
| Random access | O(1) | O(n) |

**Syntax:**
```python
deque(iterable, maxlen=None)
```

````python
# ============================================================
# DEQUE — BASICS
# ============================================================

dq = deque([1, 2, 3, 4, 5])
print("Initial deque:", dq)

# Adding elements
dq.append(6)        # Add to right
dq.appendleft(0)    # Add to left
print("After append(6) and appendleft(0):", dq)

# Adding multiple elements
dq.extend([7, 8])          # Extend right
dq.extendleft([-2, -1])    # Extend left (note: reverses order!)
print("After extend and extendleft:", dq)

# Removing elements
right = dq.pop()           # Remove from right
left  = dq.popleft()       # Remove from left
print(f"Popped right: {right}, left: {left}")
print("After pops:", dq)
````

````python
# ============================================================
# DEQUE — ROTATE (Unique Feature!)
# ============================================================

dq = deque([1, 2, 3, 4, 5])
print("Original:", dq)

dq.rotate(2)    # Rotate right by 2
print("rotate(2):", dq)

dq.rotate(-2)   # Rotate left by 2 (undo)
print("rotate(-2):", dq)

dq.rotate(1)    # Rotate right by 1
print("rotate(1):", dq)
````

````python
# ============================================================
# DEQUE — maxlen: Sliding Window / Fixed-Size Buffer
# ============================================================

# A deque with maxlen automatically discards from the opposite end
window = deque(maxlen=5)

print("Sliding window of size 5:")
sensor_data = [10, 20, 15, 30, 25, 18, 22, 35, 28, 40]

for reading in sensor_data:
    window.append(reading)
    avg = sum(window) / len(window)
    print(f"  Reading: {reading:3d}  Window: {list(window)}  Avg: {avg:.1f}")
````

````python
# ============================================================
# DEQUE — IMPLEMENTING A BFS QUEUE (ML / Graph search)
# ============================================================

def bfs(graph, start):
    """Breadth-First Search using deque as the queue."""
    visited = set()
    queue   = deque([start])
    order   = []

    while queue:
        node = queue.popleft()   # O(1) — much better than list.pop(0)!
        if node not in visited:
            visited.add(node)
            order.append(node)
            queue.extend(graph.get(node, []))

    return order

# Example graph (could represent a neural network DAG)
nn_graph = {
    "input":   ["conv1", "conv2"],
    "conv1":   ["pool1"],
    "conv2":   ["pool1"],
    "pool1":   ["fc1"],
    "fc1":     ["output"],
    "output":  []
}

traversal = bfs(nn_graph, "input")
print("BFS traversal order:", traversal)
````

# 5️⃣ `namedtuple` — Readable, Lightweight Records

## 📖 What is `namedtuple`?

`namedtuple` creates **tuple subclasses with named fields** — giving you the immutability of tuples with the **readability of attributes**.

**Syntax:**
```python
MyClass = namedtuple('MyClass', ['field1', 'field2', 'field3'])
```

| Feature | Regular Tuple | `namedtuple` |
|---------|--------------|-------------|
| Immutable | ✅ | ✅ |
| Named access | ❌ | ✅ |
| Index access | ✅ | ✅ |
| Memory efficient | ✅ | ✅ |
| Self-documenting | ❌ | ✅ |

````python
# ============================================================
# NAMEDTUPLE — BASICS
# ============================================================

# Define a namedtuple type
Point = namedtuple("Point", ["x", "y"])
Color = namedtuple("Color", "red green blue")  # can also use space-separated string

# Create instances
p = Point(3, 4)
c = Color(255, 128, 0)

# Access by name (readable!)
print(f"Point:  x={p.x}, y={p.y}")
print(f"Color:  R={c.red}, G={c.green}, B={c.blue}")

# Access by index (still works like a tuple)
print(f"Point[0]={p[0]}, Point[1]={p[1]}")

# Unpack like a regular tuple
x, y = p
print(f"Unpacked: x={x}, y={y}")

# It's still a tuple!
print(f"isinstance(p, tuple): {isinstance(p, tuple)}")
````

````python
# ============================================================
# NAMEDTUPLE — BUILT-IN METHODS
# ============================================================

Employee = namedtuple("Employee", ["name", "department", "salary", "years"])

emp = Employee("Alice", "Engineering", 95000, 5)
print("Employee:", emp)

# _asdict() → convert to OrderedDict
emp_dict = emp._asdict()
print("As dict:", emp_dict)

# _replace() → create modified copy (since immutable!)
promoted = emp._replace(salary=110000, years=6)
print("After promotion:", promoted)
print("Original unchanged:", emp)

# _fields → field names
print("Fields:", Employee._fields)

# _make() → create from iterable
data_row = ["Bob", "Marketing", 78000, 3]
bob = Employee._make(data_row)
print("Created with _make():", bob)
````

````python
# ============================================================
# NAMEDTUPLE — WITH DEFAULTS (Python 3.6.1+)
# ============================================================

# Defaults apply from RIGHT to LEFT
ModelConfig = namedtuple(
    "ModelConfig",
    ["model_type", "learning_rate", "epochs", "batch_size"],
    defaults=[0.001, 100, 32]  # last 3 fields get defaults
)

# Only model_type is required
config1 = ModelConfig("ResNet50")
config2 = ModelConfig("LSTM", learning_rate=0.01, epochs=50)
config3 = ModelConfig("Transformer", 0.0001, 200, 64)

print("Config 1:", config1)
print("Config 2:", config2)
print("Config 3:", config3)
````

````python
# ============================================================
# NAMEDTUPLE — ML EXAMPLE: Dataset Records
# ============================================================

DataPoint = namedtuple("DataPoint", ["features", "label", "weight"])

# Simulate a mini-dataset
dataset = [
    DataPoint(features=[5.1, 3.5, 1.4, 0.2], label="setosa",     weight=1.0),
    DataPoint(features=[6.2, 3.4, 5.4, 2.3], label="virginica",  weight=1.0),
    DataPoint(features=[5.9, 3.0, 4.2, 1.5], label="versicolor", weight=1.5),
    DataPoint(features=[4.6, 3.1, 1.5, 0.2], label="setosa",     weight=1.0),
]

print("Dataset:")
for i, dp in enumerate(dataset):
    print(f"  [{i}] label={dp.label:<12} features={dp.features}  weight={dp.weight}")

# Easy analysis with named access
labels = Counter(dp.label for dp in dataset)
print("\nLabel distribution:", dict(labels))

avg_weight = sum(dp.weight for dp in dataset) / len(dataset)
print(f"Average sample weight: {avg_weight:.2f}")
````

# 6️⃣ `ChainMap` — Layered Dictionary Lookups

## 📖 What is `ChainMap`?

`ChainMap` groups multiple dictionaries into a **single logical view**. Lookups search through the maps in order until the key is found.

**Think of it as:** priority-layered configuration — local overrides global.

```python
ChainMap(map1, map2, map3, ...)
# Lookup order: map1 → map2 → map3
```

````python
# ============================================================
# CHAINMAP — BASICS
# ============================================================

# Classic use case: layered config (local > user > global defaults)
global_defaults = {"debug": False, "log_level": "WARNING", "timeout": 30, "retries": 3}
user_settings   = {"log_level": "INFO", "timeout": 60}
local_overrides = {"debug": True, "timeout": 5}

# ChainMap: first map has highest priority
config = ChainMap(local_overrides, user_settings, global_defaults)

print("Effective configuration:")
for key in ["debug", "log_level", "timeout", "retries"]:
    print(f"  {key}: {config[key]}")

print("\nFull ChainMap:", config)
````

````python
# ============================================================
# CHAINMAP — WRITES GO TO THE FIRST MAP ONLY
# ============================================================

base = {"a": 1, "b": 2}
extra = {"c": 3, "d": 4}

cm = ChainMap({}, base, extra)  # Empty first map for writes
print("Before write:", dict(cm))

cm["e"] = 99      # Writes to the first (empty) map
cm["a"] = 100     # Shadows 'a' from base, doesn't modify base

print("After write:")
print("  ChainMap 'a':", cm["a"])       # 100 (from first map)
print("  base 'a':", base["a"])          # 1 (unchanged!)
print("  maps[0]:", cm.maps[0])          # only the first map

# .new_child() — create a fresh sub-scope
child_cm = cm.new_child({"x": 99})
print("\nnew_child 'x':", child_cm["x"])
print("new_child 'a':", child_cm["a"])   # inherited
````

````python
# ============================================================
# CHAINMAP — SIMULATING VARIABLE SCOPING (like Python's own scope!)
# ============================================================

# Python uses ChainMap-like layering internally for variable lookup:
# local → enclosing → global → built-in

class SimpleInterpreter:
    """A mini interpreter that uses ChainMap for variable scoping."""

    def __init__(self):
        self.scope = ChainMap()  # Start with empty global scope

    def set_global(self, name, value):
        self.scope.maps[-1][name] = value

    def enter_scope(self):
        self.scope = self.scope.new_child()

    def exit_scope(self):
        self.scope = self.scope.parents

    def assign(self, name, value):
        self.scope[name] = value

    def lookup(self, name):
        return self.scope.get(name, "UNDEFINED")


interp = SimpleInterpreter()
interp.set_global("x", 10)
interp.set_global("y", 20)

print("Global scope → x:", interp.lookup("x"))

interp.enter_scope()        # Enter function scope
interp.assign("x", 999)    # Local x shadows global x
interp.assign("z", 42)
print("Function scope → x:", interp.lookup("x"))   # 999
print("Function scope → y:", interp.lookup("y"))   # 20 (from global)
print("Function scope → z:", interp.lookup("z"))   # 42

interp.exit_scope()         # Exit function scope
print("Back to global → x:", interp.lookup("x"))   # 10 (global unchanged)
print("Back to global → z:", interp.lookup("z"))   # UNDEFINED
````

# 7️⃣ `UserDict`, `UserList`, `UserString` — Extend Built-In Collections

## 📖 What are these?

These are **wrapper classes** that allow you to easily subclass and extend `dict`, `list`, and `str` behavior without the pitfalls of directly subclassing built-in types.

> ✅ **Best practice:** Subclass `UserDict` / `UserList` / `UserString` instead of `dict` / `list` / `str` when you need custom behavior — safer and more predictable.

````python
# ============================================================
# USERDICT — A Validated Config Dictionary
# ============================================================

class TypedConfig(UserDict):
    """A config dict that validates types on assignment."""

    SCHEMA = {
        "learning_rate": float,
        "epochs":        int,
        "batch_size":    int,
        "model_name":    str,
        "use_gpu":       bool,
    }

    def __setitem__(self, key, value):
        if key in self.SCHEMA:
            expected_type = self.SCHEMA[key]
            if not isinstance(value, expected_type):
                raise TypeError(
                    f"Key '{key}' expects {expected_type.__name__}, got {type(value).__name__}"
                )
        super().__setitem__(key, value)


cfg = TypedConfig()
cfg["learning_rate"] = 0.001
cfg["epochs"]        = 100
cfg["model_name"]    = "ResNet50"
cfg["use_gpu"]       = True
print("Valid config:", cfg)

# Try to set wrong type
try:
    cfg["epochs"] = "one_hundred"   # Should raise TypeError
except TypeError as e:
    print(f"\n❌ TypeError caught: {e}")
````

````python
# ============================================================
# USERLIST — A Bounded List with Max Length
# ============================================================

class BoundedList(UserList):
    """A list that enforces a maximum number of items."""

    def __init__(self, max_size, initlist=None):
        self.max_size = max_size
        super().__init__(initlist)
        if len(self.data) > max_size:
            raise ValueError(f"Initial list exceeds max_size={max_size}")

    def append(self, item):
        if len(self.data) >= self.max_size:
            raise OverflowError(f"List is full (max_size={self.max_size})")
        super().append(item)

    def insert(self, i, item):
        if len(self.data) >= self.max_size:
            raise OverflowError(f"List is full (max_size={self.max_size})")
        super().insert(i, item)


# Use as a fixed-capacity feature buffer
feature_buffer = BoundedList(max_size=3)
feature_buffer.append("feature_A")
feature_buffer.append("feature_B")
feature_buffer.append("feature_C")
print("Buffer:", feature_buffer)

try:
    feature_buffer.append("feature_D")
except OverflowError as e:
    print(f"❌ OverflowError: {e}")
````

````python
# ============================================================
# USERSTRING — A Custom Text Processor
# ============================================================

class SanitizedString(UserString):
    """A string that auto-strips and lowercases on creation."""

    def __init__(self, seq=''):
        super().__init__(seq.strip().lower())

    def word_count(self):
        return len(self.data.split())

    def is_palindrome(self):
        clean = ''.join(c for c in self.data if c.isalnum())
        return clean == clean[::-1]


s = SanitizedString("  Hello World From Python  ")
print("Sanitized:", repr(s))
print("Word count:", s.word_count())
print("Is palindrome:", s.is_palindrome())

p = SanitizedString("  A man a plan a canal Panama  ")
print("\nSanitized:", repr(p))
print("Is palindrome:", p.is_palindrome())
````

# 8️⃣ Real-World ML Applications

Let's combine multiple `collections` tools in realistic ML scenarios.

````python
# ============================================================
# ML APPLICATION 1: Feature Engineering Pipeline
# ============================================================

# Using namedtuple for structured data, Counter for stats,
# defaultdict for grouping

Sample = namedtuple("Sample", ["id", "category", "value", "label"])

raw_data = [
    Sample(1, "tech",    4.5, 1),
    Sample(2, "health",  2.1, 0),
    Sample(3, "tech",    3.8, 1),
    Sample(4, "finance", 5.0, 1),
    Sample(5, "health",  1.5, 0),
    Sample(6, "finance", 4.2, 1),
    Sample(7, "tech",    2.9, 0),
    Sample(8, "health",  3.3, 0),
]

# 1. Label distribution (class imbalance check)
label_dist = Counter(s.label for s in raw_data)
print("Label distribution:", dict(label_dist))

# 2. Group by category
by_category = defaultdict(list)
for s in raw_data:
    by_category[s.category].append(s)

# 3. Compute per-category stats
print("\nPer-category statistics:")
print(f"  {'Category':<12} {'Count':>6}  {'Avg Value':>10}  {'Positive Rate':>14}")
print("  " + "-" * 48)
for cat in sorted(by_category):
    samples  = by_category[cat]
    count    = len(samples)
    avg_val  = sum(s.value for s in samples) / count
    pos_rate = sum(s.label for s in samples) / count
    print(f"  {cat:<12} {count:>6}  {avg_val:>10.2f}  {pos_rate:>14.1%}")
````

````python
# ============================================================
# ML APPLICATION 2: Hyperparameter Config Management
# ============================================================

# Using ChainMap for layered hyperparameter resolution
# and namedtuple for experiment tracking

ExperimentResult = namedtuple(
    "ExperimentResult",
    ["exp_id", "model", "accuracy", "loss", "config"],
    defaults=[None]
)

# Hyperparameter layers
global_hp  = {"lr": 0.001, "epochs": 100, "optimizer": "adam", "dropout": 0.3}
cnn_hp     = {"lr": 0.0005, "batch_size": 64, "kernel_size": 3}
exp_hp     = {"epochs": 50, "dropout": 0.5}   # This experiment's overrides

effective_config = ChainMap(exp_hp, cnn_hp, global_hp)

# Simulate experiment run
import random
random.seed(42)

results = deque(maxlen=5)  # Keep only last 5 experiment results

for exp_id in range(1, 8):
    # Vary LR slightly across experiments
    run_config = effective_config.new_child({"lr": effective_config["lr"] * (0.9 ** exp_id)})
    acc  = round(0.75 + random.uniform(0, 0.2), 4)
    loss = round(random.uniform(0.1, 0.5), 4)
    result = ExperimentResult(exp_id, "CNN", acc, loss, dict(run_config))
    results.append(result)

print(f"Last {results.maxlen} experiment results (rolling window):")
print(f"{'Exp':>4}  {'Model':<8}  {'Accuracy':>9}  {'Loss':>7}")
print("-" * 35)
for r in results:
    print(f"{r.exp_id:>4}  {r.model:<8}  {r.accuracy:>9.4f}  {r.loss:>7.4f}")

# Best experiment
best = max(results, key=lambda r: r.accuracy)
print(f"\nBest: Exp #{best.exp_id} with accuracy={best.accuracy}")
````

````python
# ============================================================
# ML APPLICATION 3: Confusion Matrix with defaultdict
# ============================================================

# Simulate predictions
y_true = ["cat", "dog", "cat", "bird", "dog", "cat", "bird", "dog", "cat", "bird"]
y_pred = ["cat", "cat", "cat", "bird", "dog", "dog", "bird", "dog", "cat", "cat"]

# Build confusion matrix using defaultdict
confusion = defaultdict(lambda: defaultdict(int))
for true, pred in zip(y_true, y_pred):
    confusion[true][pred] += 1

classes = sorted(set(y_true))
print("Confusion Matrix (rows=true, cols=predicted):")
print(f"{'':>8}", end="")
for cls in classes:
    print(f"  {cls:>6}", end="")
print()

for true_cls in classes:
    print(f"{true_cls:>8}", end="")
    for pred_cls in classes:
        print(f"  {confusion[true_cls][pred_cls]:>6}", end="")
    print()

# Per-class accuracy
print("\nPer-class recall:")
class_totals = Counter(y_true)
for cls in classes:
    correct = confusion[cls][cls]
    total   = class_totals[cls]
    recall  = correct / total if total > 0 else 0
    print(f"  {cls}: {correct}/{total} = {recall:.1%}")
````

# 📊 Summary — When to Use Which?

| Collection | Best For | Time Complexity Highlight |
|-----------|----------|---------------------------|
| `Counter` | Frequency analysis, histograms, NLP | `most_common()`: O(n log k) |
| `defaultdict` | Grouping, inverted indices, graphs | Same as `dict`: O(1) avg |
| `OrderedDict` | LRU cache, ordered processing | `move_to_end()`: O(1) |
| `deque` | Queues, sliding windows, BFS/DFS | `appendleft/popleft`: O(1) |
| `namedtuple` | Structured records, readable data | Same as `tuple`: O(1) access |
| `ChainMap` | Layered configs, variable scoping | Lookup: O(k) where k=maps |
| `UserDict/List/String` | Custom validated collections | Same as wrapped type |

---

## 🎓 Key Takeaways

1. **`Counter`** eliminates manual counting boilerplate and adds powerful set-like math operations.
2. **`defaultdict`** prevents `KeyError` and makes grouping/accumulating trivially easy.
3. **`OrderedDict`** provides order-sensitive equality and the essential `move_to_end()` — ideal for LRU cache.
4. **`deque`** is the go-to for O(1) queue operations; use `maxlen` for sliding windows.
5. **`namedtuple`** makes code self-documenting by giving tuple fields names — zero memory overhead.
6. **`ChainMap`** is perfect for configuration layering and mirrors Python's own scoping rules.
7. **`UserDict/List/String`** are the safe way to extend built-in containers with custom behavior.
