```python
import pandas as pd
import numpy as np

raw_data = """
ID:101 | Name:  _Rahim Ahmed_  | Price: 5000tk | Status: Active
ID:102 | Name: Karim_Hasan | Price: 12000tk | Status: Inactive
ID:103 | Name:   Tina   | Price: 8500tk | Status: Active
ID:104 | Name: Sumon_Bhai | Price: 3000tk | Status: Active
ID:105 | Name: _Missing_Data_ | Status: Active  
""" 

cleaned_records = []

for line in raw_data.strip().split('\n'):
    parts = line.split('|')
    
    record_id = parts[0].split(':')[1].strip()
    
    raw_name = parts[1].split(':')[1].strip()
    clean_name = raw_name.replace('_', ' ').strip()
    
    price_str = parts[2].split(':')[1].strip() if len(parts) > 2 else "0tk"
    
    clean_price = int(price_str[:-2]) if price_str.endswith('tk') else 0
    
    status_str = parts[3].split(':')[1].strip() if len(parts) > 3 else "Unknown"

    record = {
        "ID": record_id,
        "Name": clean_name,
        "Price": clean_price,
        "Status": status_str
    }
    cleaned_records.append(record)

print("✅ Cleaned dictionary list:")
for r in cleaned_records:
    print(r)

df = pd.DataFrame(cleaned_records).set_index("ID")
print("\n📊 Pandas DataFrame:")
print(df)

print("\n--- 4.1: Using .loc to get specific ID data ---")
print(df.loc["103", "Name":"Price"])

print("\n--- 4.2: Using .iloc for position-based data ---")
print(df.iloc[-2:, 0:2])

print("\n--- 4.3: Boolean Filtering ---")
active_high_value = df[(df["Status"] == "Active") & (df["Price"] > 4000)]
print(active_high_value)

print("\n--- 4.4: NumPy 2D Slicing Combo ---")
numpy_matrix = df[["Price", "Status"]].values
print("NumPy Array Slice (Last 3 rows, all cols):")
print(numpy_matrix[-3:, :])
```
