```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import plotly.express as px
```

```python
dict ={
  'day' : [1,2,3,4,5],
  'visit':[6,7,8,9,12]
}
df= pd.DataFrame(dict)
df.sample()

arr = np.array(df)
arr = arr*3
arr

df_2 = pd.DataFrame(arr)
df_2.columns = ["day","visit"]
df_2
```

```python
# =========================================================
# 🚀 PROFESSIONAL ML DATA VISUALIZATION MASTER GUIDE
# =========================================================

# =========================
# 📦 Imports
# =========================

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_iris
from sklearn.metrics import (
    confusion_matrix,
    roc_curve,
    auc,
    precision_recall_curve
)

from sklearn.decomposition import PCA
from sklearn.model_selection import learning_curve
from sklearn.linear_model import LogisticRegression

# =========================
# 🎨 Visualization Theme
# =========================

sns.set_theme(
    style="whitegrid",
    palette="deep"
)

# =========================
# 📦 Load Dataset
# =========================

tips = sns.load_dataset("tips")
iris = sns.load_dataset("iris")

# =========================================================
# 1️⃣ HISTOGRAM + KDE
# =========================================================

plt.figure(figsize=(8, 5))

sns.histplot(
    tips["total_bill"],
    kde=True,
    bins=20,
    color="skyblue"
)

plt.title("Distribution of Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")

plt.show()

# =========================================================
# 2️⃣ COUNTPLOT (CLASS IMBALANCE)
# =========================================================

plt.figure(figsize=(7, 5))

sns.countplot(
    x="day",
    data=tips,
    hue="day",
    legend=False
)

plt.title("Count of Records Per Day")

plt.show()

# =========================================================
# 3️⃣ CORRELATION HEATMAP
# =========================================================

corr = tips.select_dtypes(include=np.number).corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()

# =========================================================
# 4️⃣ BOXPLOT
# =========================================================

plt.figure(figsize=(8, 5))

sns.boxplot(
    x="day",
    y="total_bill",
    hue="sex",
    data=tips
)

plt.title("Boxplot: Bill Distribution")

plt.show()

# =========================================================
# 5️⃣ VIOLIN PLOT
# =========================================================

plt.figure(figsize=(8, 5))

sns.violinplot(
    x="day",
    y="total_bill",
    data=tips
)

plt.title("Violin Plot: Distribution Shape")

plt.show()

# =========================================================
# 6️⃣ SCATTERPLOT
# =========================================================

plt.figure(figsize=(8, 5))

sns.scatterplot(
    x="total_bill",
    y="tip",
    hue="time",
    style="time",
    data=tips,
    s=100,
    alpha=0.7
)

plt.title("Scatterplot: Bill vs Tip")

plt.show()

# =========================================================
# 7️⃣ PAIRPLOT (FAST EDA)
# =========================================================

sns.pairplot(
    iris,
    hue="species"
)

plt.show()

# =========================================================
# 8️⃣ JOINTPLOT
# =========================================================

sns.jointplot(
    data=tips,
    x="total_bill",
    y="tip",
    kind="hex"
)

plt.show()

# =========================================================
# 9️⃣ MISSING VALUE HEATMAP
# =========================================================

# Create fake missing values
tips_missing = tips.copy()

tips_missing.loc[0:20, "total_bill"] = np.nan

plt.figure(figsize=(8, 4))

sns.heatmap(
    tips_missing.isnull(),
    cbar=False,
    cmap="viridis"
)

plt.title("Missing Value Heatmap")

plt.show()

# =========================================================
# 🔟 CONFUSION MATRIX
# =========================================================

y_true = [0, 1, 0, 1, 0, 1, 1, 0]
y_pred = [0, 1, 1, 1, 0, 0, 1, 0]

cm = confusion_matrix(y_true, y_pred)

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()

# =========================================================
# 1️⃣1️⃣ ROC CURVE
# =========================================================

y_scores = np.array([
    0.1, 0.4, 0.35, 0.8,
    0.1, 0.1, 0.9, 0.2
])

fpr, tpr, thresholds = roc_curve(
    y_true,
    y_scores
)

roc_auc = auc(fpr, tpr)

plt.figure(figsize=(7, 5))

plt.plot(
    fpr,
    tpr,
    label=f"AUC = {roc_auc:.2f}"
)

plt.plot([0, 1], [0, 1], linestyle="--")

plt.title("ROC Curve")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")

plt.legend()

plt.show()

# =========================================================
# 1️⃣2️⃣ PRECISION-RECALL CURVE
# =========================================================

precision, recall, _ = precision_recall_curve(
    y_true,
    y_scores
)

plt.figure(figsize=(7, 5))

plt.plot(recall, precision)

plt.title("Precision-Recall Curve")
plt.xlabel("Recall")
plt.ylabel("Precision")

plt.show()

# =========================================================
# 1️⃣3️⃣ FEATURE IMPORTANCE
# =========================================================

features = [
    "Total Bill",
    "Size",
    "Day_Sun",
    "Time_Dinner",
    "Smoker_Yes"
]

importances = [0.45, 0.25, 0.15, 0.10, 0.05]

importance_df = pd.DataFrame({
    "Feature": features,
    "Importance": importances
})

plt.figure(figsize=(8, 5))

sns.barplot(
    data=importance_df,
    x="Importance",
    y="Feature",
    hue="Feature",
    palette="viridis",
    legend=False
)

plt.title("Feature Importance")

plt.show()

# =========================================================
# 1️⃣4️⃣ PCA VISUALIZATION
# =========================================================

X = iris.drop("species", axis=1)

pca = PCA(n_components=2)

X_pca = pca.fit_transform(X)

pca_df = pd.DataFrame({
    "PCA1": X_pca[:, 0],
    "PCA2": X_pca[:, 1],
    "species": iris["species"]
})

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=pca_df,
    x="PCA1",
    y="PCA2",
    hue="species",
    s=100
)

plt.title("PCA Visualization")

plt.show()

# =========================================================
# 1️⃣5️⃣ LEARNING CURVE
# =========================================================

X = iris.drop("species", axis=1)
y = iris["species"]

model = LogisticRegression(max_iter=200)

train_sizes, train_scores, test_scores = learning_curve(
    model,
    X,
    y,
    cv=5
)

train_mean = train_scores.mean(axis=1)
test_mean = test_scores.mean(axis=1)

plt.figure(figsize=(8, 5))

plt.plot(train_sizes, train_mean, label="Training Score")
plt.plot(train_sizes, test_mean, label="Validation Score")

plt.title("Learning Curve")
plt.xlabel("Training Examples")
plt.ylabel("Score")

plt.legend()

plt.show()

# =========================================================
# ✅ END
# =========================================================
```
