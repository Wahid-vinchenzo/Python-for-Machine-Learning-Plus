Problem 2

Write a Python generator function called batch_generator. It should take a list of numbers (representing data samples) and a batch_size. The generator should yield one batch at a time as a list.


Example Input:

data = [1, 2, 3, 4, 5, 6, 7, 8]

batch_size = 3


Expected Output (when iterated):


[1, 2, 3]

[4, 5, 6]

[7, 8]







```python
def batch_generator(data, batch_size):
    for i in range(0, len(data), batch_size):
        yield data[i:i + batch_size]

batch_size = int(input())
data = list(map(int, input().split()))

for batch in batch_generator(data, batch_size):
    print(batch)
```

Problem 3

 Given a list [1,2,3,4], use map() and lambda to create a new list with squares of each number.

Sample Input:
 No input

Sample Output:
[1, 4, 9, 16]






```python
l=[1,2,3,4]
square_list = list(map(lambda x: x**2, l))
print(square_list)
```

Problem 4

 Write a function power(base, exponent) that returns the result of base raised to exponent. If the exponent is not given, it should calculate the square.


Sample Input:

5
Sample Output:

25

Sample Input 2:

5 3

Sample Output 2:


125







```python
def power(base, exponent=None):
    if exponent is None:
        return base ** 2
    else:
        return base ** exponent

user_input = input().split()

if len(user_input) == 1:
    base = int(user_input[0])
    exponent = None
elif len(user_input) == 2:
    base = int(user_input[0])
    exponent = int(user_input[1])
else:
    print("Please provide one or two numbers (base and optional exponent).")

    base = 0
    exponent = None

result = power(base, exponent)
print(result)
```

Problem 5 ( Conceptual Session )

 Given a list [1,2,3,4,5,6], use filter() to select even numbers and map() with lambda to square them.

Sample Input:

 No input

Sample Output:

[4, 16, 36]







```python
l=[1,2,3,4,5,6]

en = list(filter(lambda x: x % 2 == 0, l))
sen = list(map(lambda x: x**2, en))

print(sen)
```
